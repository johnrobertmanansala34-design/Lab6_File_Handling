```python
import csv
import json
import os
import shutil

SEED = 9
TARGET_OS = "iOS" if SEED % 2 != 0 else "Android"
TARGET_BEHAVIOR_CLASS = (SEED % 5) + 1

input_file = "user_behavior_dataset.csv"
output_dir = "output"
os.makedirs(output_dir, exist_ok=True)

exported_csv_path = os.path.join(output_dir, "filtered_device_usage.csv")
json_summary_path = os.path.join(output_dir, "device_usage_summary.json")
backup_csv_path = os.path.join(output_dir, "filtered_device_usage_backup.csv")

total_records = 0
matching_os_count = 0
matching_behavior_count = 0
total_battery_drain = 0.0
total_data_usage = 0.0
exported_rows = []

print(
    f"Processing dataset for OS: {TARGET_OS} and Behavior Class:"
    f" {TARGET_BEHAVIOR_CLASS}..."
)
```

```python
with open(input_file, mode="r", encoding="utf-8") as f:
    reader = csv.reader(f)
    header = next(reader, None)

    for row in reader:
        total_records += 1
        try:
            user_id = row[0].strip()
            device_model = row[1].strip()
            os_type = row[2].strip()
            battery_drain = float(row[5].strip())
            data_usage = float(row[7].strip())
            behavior_class = int(row[10].strip())

            if os_type.lower() == TARGET_OS.lower():
                matching_os_count += 1

                if behavior_class == TARGET_BEHAVIOR_CLASS:
                    matching_behavior_count += 1
                    total_battery_drain += battery_drain
                    total_data_usage += data_usage

                    exported_rows.append(
                        [user_id, device_model, os_type, battery_drain, data_usage]
                    )
        except (IndexError, ValueError):
            continue

    if matching_behavior_count > 0:
        avg_battery_drain = total_battery_drain / matching_behavior_count
        avg_data_usage = total_data_usage / matching_behavior_count
    else:
        avg_battery_drain = 0.0
        avg_data_usage = 0.0

    with open(exported_csv_path, mode="w", newline="", encoding="utf-8") as f_out:
        writer = csv.writer(f_out)
        writer.writerow([
            "User ID",
            "Device Model",
            "Operating System",
            "Battery Drain (mAh/day)"
            "Data Usage (MB/day)"
        ])
        writer.writerows(exported_rows)

    summary_data = {
        "operating_system": TARGET_OS,
        "behavior_class": TARGET_BEHAVIOR_CLASS,
        "total_records_in_dataset": total_records,
        "records_matching_os": matching_os_count,
        "records_in_target_behavior_class": matching_behavior_count,
        "average_battery_drain_mah": round(avg_battery_drain, 2),
        "average_data_usage_mb": round(avg_data_usage, 2),
    }

    with open(json_summary_path, mode="w", encoding="utf-8") as j_out:
        json.dump(summary_data, j_out, indent=4)

    shutil.copy(exported_csv_path, backup_csv_path)
    backup_size_bytes = os.path.getsize(backup_csv_path)
```

```python
print("\n--- Processing Complete ---")
print(f"Operating system and behavior class: {TARGET_OS} and {TARGET_BEHAVIOR_CLASS}")
print(f"Total records in dataset: {total_records}")
print(f"Records matching operating system: {matching_os_count}")
print(f"Records in target behavior class: {matching_behavior_count}")
print(f"Average battery drain (mAh/day): {avg_battery_drain:.2f}")
print(f"Average data usage (MB/day): {avg_data_usage:.2f}")
print(f"Count verification: {exported_rows} Match Verified")
print(f"Exported CSV file: {exported_csv_path}")
print(f"JSON summary file: {json_summary_path}")
print(f"Backup file and size: {backup_csv_path}")
```
